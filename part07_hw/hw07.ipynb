### 데이터 전처리 및 가공 2 - 결측치

```python
mpg.loc[[64, 123, 130, 152, 211], 'hwy'] = np.nan
```

```python
pd.isna(mpg['drv']).sum()
```

```python
pd.isna(mpg['hwy']).sum()
```

##### 평균 비교

```python
mpg['drv'].unique()
```

drv의 종류를 먼저 확인했다.


```python
mpg.dropna(subset=['hwy']).groupby('drv')['hwy'].mean()
```

하나의 pandas 구문을 통해 hwy의 평균을 비교해봤다. 그 결과 f 인 전륜구동의 평균이 가장 높은 것을 확인할 수 있다.

### 데이터 전처리 및 가공 2- 이상치

```python
import pandas as pd
import numpy as np
mpg = pd.read_csv('mpg.csv')
```

```python
mpg.loc[[9, 13, 57, 92], 'drv'] = 'k'
mpg.loc[[28, 42, 128, 202], 'cty'] = [3, 4, 39, 42]
```

##### drv 이상치 확인과 결측 처리

```python
mpg['drv'].value_counts().sort_index()
```

빈도표를 만들어보면, 데이터에 이상치가 있는지 확인할 수 있다. drv 변수의 정상적인 값이 4, f, r 이므로 k 라는 이상치가 있다는 것을 알 수 있다.

결측 처리를 한 후 이상치가 사라졌는지 확인해보자.

```python
mpg.loc[~mpg['drv'].isin(['4', 'f', 'r']), 'drv'] = np.nan
```

```python
mpg['drv'].value_counts().sort_index()
```

이상치가 사라졌음을 알 수 있다.
