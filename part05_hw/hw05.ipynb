```python
import pandas as pd
mpg = pd.read_csv('mpg.csv')
mpg
```

### 1. displ 4 이하인 자동차와 5 이상인 자동차 hwy의 평균 비교

```python
displ_1 = mpg.query('displ <= 4')
```

```python
displ_2 = mpg.query('displ >= 5')
```

```python
displ_1['hwy'].mean()
```

```python
displ_2['hwy'].mean()
```

displ_1['hwy'].mean()가 displ_2['hwy'].mean() 보다 높다.
따라서, 배기량이 4 이하인 자동차의 고속도로 연비 평균이 더 높다.

### 2. audi 와 toyota 중 어느 자동차 제조 회사의 도시 연비 평균이 더 높은가

```python
audi_m = mpg.query('manufacturer == "audi"')
```

```python
toyota_m = mpg.query('manufacturer == "toyota"')
```

```python
audi_m['cty'].mean()
```

```python
toyota_m['cty'].mean()
```

toyota_m['cty'].mean()가 audi_m['cty'].mean()보다 높다. 따라서 toyota의 도시 연비 평균이 더 높다.
