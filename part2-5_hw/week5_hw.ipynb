### [] 이용하기

```python
import pandas as pd
df_raw = pd.read_csv('exam.csv')
df=df_raw.head(10)
df
```

```python
df['nclass']
```

```python
df['nclass']==1
```

논리 연산자를 사용하면 조건 여부를 True / False로 확인할 수 있음

```python
df[df['nclass']==1]
```

```python
df[df['front']>=80]
```

```python
df[(df['nclass'])==1 &(df['front']>=50)]
```

```python
df[(df['front']<50) | (df['back'] < 50)]
```

여러 조건을 동시에 충족하고 싶으면 & 사용,
하나 이상 충족하는 행을 추출하려면 | 사용

```python
df['id']
```

```python
df['nclass']
```

```python
df[['id', 'nclass']]
```

[] 하고 변수명 지정하면 열 추출 가능!

```python
df['id']
```

```python
df[['id']]
```

변수를 1개만 추출하면서 데이터프레임 자료 구조를 유지하려면 변수명을 []로 한 번 더 감싸면 된다.

열 1개만 추출 : 시리즈 자료 구조로 바뀜

열 2개 이상 추출하면 df 구조 유지

```python
df[df['nclass']==1]['front']
```

```python
df[df['nclass']==1][['front', 'back']]
```

여러 열을 추출하려면 [] 를 이용해서 변수명 나열

```python
df.groupby('nclass').agg(front=('front', 'mean'), back =('back', 'mean'))
```

```python
df.groupby('nclass')[['front', 'back']].mean()
```
