### df.loc[] 이용하기

```python
df = pd.DataFrame({'var1': [1,2,3], 'var2' : [4,5,6]})
df
```

```python
df = pd.DataFrame({'var1': [1,2,3], 'var2' : [4,5,6]}, index=['kim', 'lee', 'park'])
df
```

위 예제를 통해 index 를 이용하면 인덱스 문자열을 지정할 수 있다는 걸 확인할 수 있음

```python
df.loc['kim']
```

```python
df.loc[['kim', 'park']]
```

```python
df.loc['lee', 'var1']
```

```python
df.loc[['kim', 'lee'], 'var2']
```

```python
#df.loc[, 'var1']
```

위의 코드는 에러가 발생함!!

[]에 쉼표를 입력하고 그 앞에는 행을 가리키는 인덱스 문자열, 뒤에는 열을 가리키는 변수명을 입력한다.

df.loc[]를 이용해 열을 추출할 때 반드시 행 추출 기준도 같이 입력해야 함

```python
df.loc[:, 'var1']
```

모든 행을 추출하려면 :를 입력

```python
df = df_raw.copy()
df
```

```python
df.loc[0]
```

```python
df.loc[[0]]
```

```python
df.loc[[0,3,5]]
```

```python
df.loc[0:3]
```

슬라이싱 : ':'를 이용해 연속된 행을 추출하는 작업

1. : 앞에 값 입력 안 하면 첫번째 행부터 추출

2. : 뒤에 값 입력 안 하면 마지막 행까지 추출

3. : 앞뒤에 아무 값도 입력하지 않으면 모든 행 추출

```python
df.loc[2, :]
```

```python
# df.loc[:, 2]
```

df.loc[]은 열 추출 불가. 행 추출만 가능

```python
df.loc[df['nclass']==1]
```

```python
df.loc[df['nclass']==1, ['front', 'back']]
```

```python
df.loc[0:3, 'front']
```

df.loc에 쉼표를 이용하면 조건을 충족하는 행에서 열을 추출할 수 있음

```python
# 프론트 점수 50점 초과하는 행 추출
df.query('front > 50')
```

```python
df[df['front']>50]
```

```python
df.loc[df['front']>50]
```

### df.iloc[] 이용하기

쉼표를 기준으로 앞에는 행, 뒤에는 열 기준으로 입력

인덱스 문자열 말고 인덱스 번호를 입력한다.

```python
df.iloc[0]
```

```python
df.iloc[[0]]
```

```python
df.iloc[[0,3,5]]
```

```python
df.loc[1:5]
```

```python
df.iloc[1:5]
```

df.loc[x : y] = x행  이상 y행 이하 추출 

df.iloc[x : y] = x행 이상 y행 미만 추출

```python
df_label = pd.DataFrame({'var1' : [1,2,3], 'var2' : [4,5,6]}, index=['kim', 'lee', 'park'])
df_label
```

```python
# df_label.loc[0]
```

```python
df_label.iloc[0]
```

df.iloc는 인덱스 번호로 행 추출 가능

```python
df.iloc[:, [1,3]]
```

```python
df.iloc[[0,1], [2,3]]
```

```python
# df.iloc[df['nclass]==1]
```

df.iloc[]는 조건을 지정해서 행을 추출할 수 없음

행 추출 인덱스 : df.loc[]는 문자열과 번호 <-> df.iloc[]는 번호

열 추출 인덱스 : df.loc[]는 문자열 <-> df.iloc[]는 번호

조건 지정해서 행 추출 : df.loc[]는 가능 <-> df.iloc[]는 불가능

연속된 행 추출 x:y 의미 : df.loc[]는 이상&이하 <-> df.iloc[]는 이상&미만
