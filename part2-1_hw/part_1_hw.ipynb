### 1. suv 대상 cty 평균이 가장 높은 회사 다섯 곳 막대 그래프로 표현하기

#### 평균 찾기

```python
import pandas as pd
mpg1 = pd.read_csv('mpg.csv')
```

```python
mpg1.head()
```

```python
mpg1['category'].unique()
```

```python
df_mpg1 = mpg1[mpg1['category'] == 'suv']
```

```python
df_mpg1.head()
```

```python
df_mpg1['category'].unique()
```

```python
df_mpg1 = df_mpg1.groupby('manufacturer', as_index=False).agg(mean_cty = ('cty', 'mean'))
```

```python
df_mpg1
```

```python
df_mpg1 = df_mpg1.sort_values('mean_cty', ascending=False)
df_mpg1
```

```python
import seaborn as sns
sns.barplot(data = df_mpg1[:5], x ='manufacturer', y='mean_cty')
```
